```python
import matplotlib.pyplot as plt
import numpy as np
fruit_sale = [100,200,300,400]
month = ["Jan", "Feb", "March", "April"]
plt.figure(figsize=(8,6))
plt.bar(month, fruit_sale, color = ["darkgreen", "purple", "pink","brown"], hatch= "/")
plt.title("Fruit Sale")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.show()
```

```python
study_hous =[1,2,3,4,5]
grade = [100,200,300,400,500]
plt.scatter(study_hous, grade, color = "magenta", alpha = 0.4)
plt.title("Study time")
plt.xlabel("hours")
plt.ylabel("grade")
plt.show()
```

```python
labels = ["apple", "samsung", "google"]
sales = (37, 20, 57)
plt.figure(figsize = (6,6))
plt.pie(sales, labels = labels, autopct = "%1.1f%%", shadow=True)
plt.title("Sales")
plt.show()
```

```python
month = ["Jan", "Feb", "March", "April", "May", "June","July","August", "Sept", "Oct", "Nov", "Dec"]
sales_24 = [50,53, 50, 60, 23, 70, 74, 80, 82, 91, 43,55]
sales_25 = []
for i in sales_24:
  sales_25.append(i+10)

plt.plot(month, sales_24, label = "24", color = "blue", marker="o")
plt.plot(month, sales_25, label = "25", color = "magenta", marker="s")
plt.legend()
plt.title("Sales")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.xticks(rotation = 45, ha='right')
plt.show()

```
